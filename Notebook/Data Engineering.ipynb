### Project Introduction

##### **FinShield AI: An Intelligent Financial Risk Assessment and Investment Decision Support Platform Using Machine Learning and Explainable AI**

In today's rapidly evolving financial landscape, individuals, financial institutions, and investors face increasing challenges in accurately assessing financial risk and making informed investment decisions. Traditional risk assessment methods often rely on manual analysis, historical data interpretation and expert judgment, which can be time-consuming, prone to human bias and unable to adapt quickly to dynamic market conditions. FinShield AI is an intelligent, data-driven financial analytics platform designed to address these challenges by combining Machine Learning (ML) with Explainable Artificial Intelligence (XAI). The platform leverages historical financial data, customer profiles, credit information and investment-related parameters to predict an individual's financial risk level and provide personalized investment recommendations. By integrating advanced classification algorithms with explainability techniques such as SHAP (SHapley Additive Explanations), FinShield AI enables users to understand not only the predicted risk but also the key factors influencing that prediction.

The platform follows a complete end-to-end data science workflow, including data collection, preprocessing, exploratory data analysis (EDA), feature engineering, model development, evaluation and deployment. Multiple machine learning algorithms are trained and compared to identify the most accurate and reliable model for financial risk prediction. The selected model is integrated into an interactive web application that allows users to input financial information and receive real-time risk assessments along with investment guidance.

To enhance accessibility and user experience, FinShield AI also incorporates an AI-powered financial assistant (FinShield Copilot) capable of answering finance-related queries, explaining prediction results, and assisting users in understanding investment strategies and risk management concepts. Interactive dashboards and visualizations further enable users to monitor financial insights, analyze trends, and make data-driven decisions.

The primary objective of FinShield AI is to bridge the gap between artificial intelligence and financial decision-making by delivering accurate, transparent, and explainable predictions. The platform empowers individuals, financial advisors, and organizations with actionable insights, reduces uncertainty in financial planning, and promotes responsible investment practices through interpretable AI.

By integrating Machine Learning, Explainable AI, Data Analytics, Interactive Dashboards and an AI Chatbot into a unified platform, FinShield AI demonstrates how modern AI technologies can transform financial risk assessment into a more intelligent, transparent, and trustworthy process. It serves as a comprehensive solution for enhancing financial decision-making while showcasing practical applications of AI in the FinTech domain.

### Proposed Solution

The proposed solution is the development of FinShield AI Assistant, an integrated financial intelligence platform that combines Machine Learning, Explainable AI, fraud analytics, and a conversational AI assistant to provide comprehensive financial risk analysis and decision support.

The solution follows an end-to-end data science pipeline consisting of:

##### Step 1: Data Collection

Collect financial datasets containing customer demographics, credit history, income, expenses, loan information, market price data, macroeconomic indicators, and financial news, sourced from the Home Credit Default Risk dataset and supplementary market/economic/news data.

##### Step 2: Data Preprocessing

Perform data cleaning, handle missing values, remove duplicates, encode categorical variables, and load the cleaned tables into a structured SQLite database for downstream analysis.

##### Step 3: Exploratory Data Analysis (EDA)

Analyze financial patterns, identify relationships among demographic, credit, and default variables, detect outliers, and generate visualizations such as correlation heatmaps and income-vs-expense distributions.

##### Step 4: Feature Engineering

Create meaningful financial indicators such as debt-to-income ratio, credit utilization, repayment behavior, and transaction-based behavioral ratios to improve predictive performance.

##### Step 5: Machine Learning Model Development

Train and evaluate multiple supervised and unsupervised learning algorithms, including:

- Logistic Regression (risk classification)
- Random Forest (risk classification)
- Isolation Forest (fraud/anomaly detection)
- ARIMA (market movement forecasting)
- TF-IDF classifier (news sentiment analysis)

The best-performing risk classifier is selected based on accuracy, macro F1, and ROC-AUC.

##### Step 6: Explainable AI

Implement SHAP (SHapley Additive Explanations) to provide feature-level explanations for every risk prediction, improving transparency and analyst confidence in the model's decisions.

##### Step 7: FinShield Scoring & Recommendation Engine

Blend credit, market, economic, and sentiment risk into a single composite FinShield Score (0–1000), assign a Financial Health tier, and generate rule-based recommended actions aligned with each customer's risk profile from standard monitoring to critical intervention.

##### Step 8: AI Financial Assistant

Integrate the FinShield AI Chatbot, an assistant capable of:

- Explaining a customer's FinShield Score and risk classification
- Answering natural-language questions about alert levels and fraud flags
- Surfacing recommended actions for a given customer
- Assisting analysts in interpreting portfolio-level statistics

##### Step 9: Risk Monitoring & Reporting

Produce structured outputs displaying:

- Risk tier distribution across the customer portfolio
- Individual customer financial profiles
- Model performance metrics
- SHAP feature importance
- Fraud alerts and critical-risk flags
- Segment-level financial trends

##### Step 10: Deployment

Deploy the chatbot layer as a lightweight web application using Streamlit, reading directly from the pipeline's saved outputs, enabling real-time, natural-language customer risk lookups without requiring an external API or live model inference call.

##### Step 3: Load Customer Dataset

```python
print("Loading real Home Credit data as the customer behavioral backbone.")
df = pd.read_csv(r"C:\Users\udays\Downloads\FinShield AI\application_train.csv")
df = df.sample(n=60000, random_state=42).reset_index(drop=True)
df
```

##### Step 4: Building the Customer Profile table

```python
customer_profile = pd.DataFrame({
    "customer_id": ["C" + str(10000 + i) for i in range(len(df))],
    "sk_id_curr": df["SK_ID_CURR"],
    "age": (-df["DAYS_BIRTH"] / 365.25).round().astype(int),
    "gender": df["CODE_GENDER"].replace({"XNA": "F"}),
    "occupation": df["OCCUPATION_TYPE"].fillna("Not Specified"),
    "education": df["NAME_EDUCATION_TYPE"],
    "city_tier": np.random.choice(["Tier 1", "Tier 2", "Tier 3"],len(df),p=[0.45, 0.35, 0.20]),
    "employment_years": np.where(df["DAYS_EMPLOYED"] == 365243,0,(-df["DAYS_EMPLOYED"] / 365.25).clip(lower=0)).round(1),
    "family_status": df["NAME_FAMILY_STATUS"],
    "num_children": df["CNT_CHILDREN"],})

customer_profile.to_csv("Data/raw/customer_profile.csv", index=False)
customer_profile.head(3)
```

##### Step 5: Building the Financial Transactions table

```python
financial_transactions = pd.DataFrame({
    "customer_id": customer_profile["customer_id"],
    "monthly_income": (df["AMT_INCOME_TOTAL"] / 12).round(0),
    "monthly_expense": (df["AMT_INCOME_TOTAL"] / 12 * np.random.uniform(0.35, 0.75, len(df))).round(0),
    "savings_amount": (df["AMT_INCOME_TOTAL"]* np.random.uniform(0.1, 0.9, len(df))).round(0),
    "investment_amount": (df["AMT_INCOME_TOTAL"]* np.random.uniform(0, 0.4, len(df)) * (np.random.rand(len(df)) > 0.4)).round(0),
    "loan_amount": df["AMT_CREDIT"].round(0),
    "emi_amount": df["AMT_ANNUITY"].fillna(df["AMT_ANNUITY"].median()).round(0),
    "credit_usage_pct": np.random.beta(2, 3, len(df)).round(3) * 100,
    "transaction_count_monthly": np.random.poisson(28, len(df)),})

financial_transactions.to_csv("Data/raw/financial_transactions.csv", index=False)
financial_transactions.head(3)
```

##### Step 6: Building the Credit History  table

```python
ext_mean = df[["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"]].mean(axis=1)
ext_mean_filled = ext_mean.fillna(ext_mean.median())
credit_score = (300 + ext_mean_filled * 550).round().clip(300, 900).astype(int)

credit_history = pd.DataFrame({
    "customer_id": customer_profile["customer_id"],
    "credit_score": credit_score,
    "previous_loans": np.random.poisson(2.2, len(df)),
    "late_payments": (df["DEF_30_CNT_SOCIAL_CIRCLE"].fillna(0)+ np.random.poisson(1, len(df))).clip(0, 20).astype(int),
    "loan_defaults": df["TARGET"],
    "credit_card_count": np.random.poisson(1.5, len(df)),
    "loan_duration_months": np.random.choice([12, 24, 36, 48, 60, 84, 120],len(df),p=[0.10, 0.15, 0.20, 0.20, 0.15, 0.12, 0.08]),})

credit_history.to_csv("Data/raw/credit_history.csv", index=False)
credit_history.head(3)
```

##### Step 7: Generating Synthetic Market Data

```python
companies = ["ALPHA", "BRAVO", "CIVIC", "DELTA", "ECHO", "FOXTEL", "GRANITE", "HORIZON", "INDUS", "JOVIAN"]

dates = pd.bdate_range("2024-01-01", "2025-12-31")

rows = []

for comp in companies:
    price = np.random.uniform(80, 400)
    drift = np.random.uniform(-0.0003, 0.0006)
    vol = np.random.uniform(0.012, 0.03)

    for d in dates:
        ret = np.random.normal(drift, vol)
        price = max(price * (1 + ret), 1)
        rows.append([comp,d.date().isoformat(),round(price, 2), round(ret, 5)])

market_data = pd.DataFrame(rows, columns=["company", "date", "close", "returns"])
market_data["volatility_20d"] = (market_data.groupby("company")["returns"].transform(lambda s: s.rolling(20).std()))

market_data.to_csv("Data/raw/market_data.csv", index=False)
market_data.head(3)
```

##### Step 8: Generating Synthetic Economic Indicators

```python
months = pd.date_range("2024-01-01", "2025-12-01", freq="MS")

economic_indicators = pd.DataFrame({
    "month": months.strftime("%Y-%m"),
    "gdp_growth_pct": (6.0 + np.cumsum(np.random.normal(0, 0.15, len(months)))).round(2),
    "inflation_rate_pct": (5.0 + np.cumsum(np.random.normal(0, 0.10, len(months)))).round(2),
    "interest_rate_pct": (6.5 + np.cumsum(np.random.normal(0, 0.05, len(months)))).round(2),
    "unemployment_rate_pct": (7.5 + np.cumsum(np.random.normal(0, 0.10, len(months)))).clip(3, 12).round(2),})

economic_indicators.to_csv("Data/raw/economic_indicators.csv", index=False)
economic_indicators.head(3)
```

##### Step 9: Generating Synthetic News Headlines

```python
pos = ["{c} reports record quarterly profit growth","{c} stock surges after strong earnings beat"]

neg = [ "{c} shares fall after profit warning", "Fraud investigation launched against {c} executives"]

neu = ["{c} to release quarterly results next week","{c} holds annual shareholder meeting"]

news_rows = []

for comp in companies:
    for _ in range(40):
        r = np.random.rand()
        if r < 0.35:
            text, label = np.random.choice(pos).format(c=comp), "Positive"
        elif r < 0.65:
            text, label = np.random.choice(neg).format(c=comp), "Negative"
        else:
            text, label = np.random.choice(neu).format(c=comp), "Neutral"
        news_rows.append([text, comp, label])

news_sentiment = pd.DataFrame(news_rows,columns=["headline", "company", "sentiment_label"])

news_sentiment.to_csv("Data/raw/news_sentiment.csv", index=False)
news_sentiment.head(3)
```

#### Phase 2: Cleaning the Data + Building the SQL Database

##### Step 1: Loading Raw Tables

```python
customer_profile = pd.read_csv('Data/raw/customer_profile.csv')
financial_transactions = pd.read_csv('Data/raw/financial_transactions.csv')
credit_history = pd.read_csv('Data/raw/credit_history.csv')
market_data = pd.read_csv('Data/raw/market_data.csv')
economic_indicators = pd.read_csv('Data/raw/economic_indicators.csv')
news_sentiment = pd.read_csv('Data/raw/news_sentiment.csv')
print("All 6 raw tables loaded.")
```

##### Step 2: Handling Missing Values & Capping Extreme Outliers

```python
for col in ["monthly_income", "monthly_expense", "savings_amount", "emi_amount"]:
    financial_transactions[col] = (financial_transactions[col].fillna(financial_transactions[col].median()))

financial_transactions["investment_amount"] = (financial_transactions["investment_amount"].fillna(0))

for col in ["monthly_income", "loan_amount", "emi_amount"]:
    q1, q3 = financial_transactions[col].quantile([0.01, 0.99])
    financial_transactions[col] = financial_transactions[col].clip(q1, q3)

print("Missing values filled, outliers capped at the 1st/99th percentile.")
```

##### Step 3: Saving Cleaned Data

```python
customer_profile.to_csv("Data/processed/customer_profile_clean.csv",index=False)
financial_transactions.to_csv("Data/processed/financial_transactions_clean.csv",index=False)
credit_history.to_csv("Data/processed/credit_history_clean.csv",index=False)
market_data.to_csv("Data/processed/market_data_clean.csv",index=False)
economic_indicators.to_csv("Data/processed/economic_indicators_clean.csv",index=False)
news_sentiment.to_csv("Data/processed/news_sentiment_clean.csv",index=False)
print("Cleaned files saved to Data/processed/")
```

#### Phase 3: Building the SQL Database

##### Step 1: Creating Database Tables

```python
DB_PATH = "SQL/FinShield_DB.sqlite"

if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)

conn.executescript("""
CREATE TABLE customers (
    customer_id TEXT PRIMARY KEY,
    sk_id_curr INTEGER,
    age INTEGER,
    gender TEXT,
    occupation TEXT,
    education TEXT,
    city_tier TEXT,
    employment_years REAL,
    family_status TEXT,
    num_children INTEGER
);

CREATE TABLE financial_transactions (
    customer_id TEXT,
    monthly_income REAL,
    monthly_expense REAL,
    savings_amount REAL,
    investment_amount REAL,
    loan_amount REAL,
    emi_amount REAL,
    credit_usage_pct REAL,
    transaction_count_monthly INTEGER
);

CREATE TABLE credit_history (
    customer_id TEXT,
    credit_score INTEGER,
    previous_loans INTEGER,
    late_payments INTEGER,
    loan_defaults INTEGER,
    credit_card_count INTEGER,
    loan_duration_months INTEGER
);

CREATE TABLE market_data (
    company TEXT,
    date TEXT,
    close REAL,
    returns REAL,
    volatility_20d REAL
);

CREATE TABLE economic_indicators (
    month TEXT PRIMARY KEY,
    gdp_growth_pct REAL,
    inflation_rate_pct REAL,
    interest_rate_pct REAL,
    unemployment_rate_pct REAL
);

CREATE TABLE news_sentiment (
    headline TEXT,
    company TEXT,
    sentiment_label TEXT
);

CREATE TABLE risk_scores (
    customer_id TEXT,
    credit_risk_score REAL,
    market_risk_score REAL,
    economic_risk_score REAL,
    sentiment_risk_score REAL,
    finshield_score REAL,
    risk_category TEXT
);

CREATE TABLE predictions (
    customer_id TEXT,
    model_name TEXT,
    predicted_probability REAL,
    predicted_class TEXT,
    prediction_date TEXT
);
""")

print("8 tables created.")
```

##### Step 2: Loading Cleaned Data into SQL

```python
customer_profile.to_sql("customers",conn,if_exists="append",index=False)
financial_transactions.to_sql("financial_transactions",conn,if_exists="append",index=False)
credit_history.to_sql("credit_history",conn,if_exists="append",index=False)
market_data.to_sql("market_data",conn,if_exists="append",index=False)
economic_indicators.to_sql("economic_indicators",conn,if_exists="append",index=False)
news_sentiment.to_sql("news_sentiment",conn,if_exists="append",index=False)
conn.commit()
print("Data loaded")
```
